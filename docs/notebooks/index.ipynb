## Working with files

> **The sidebar.** The left-hand **sidebar** lists the local files, which are kept in a virtual file system within the browser (managed via IndexedDB).
>
> * To create a new notebook use *File/New/Notebook*
> * To upload a notebook or data file use the upload icon above the sidebar, or drag and drop files onto the sidebar
> * To load and open a file from a URL, use *File/Open From URL...*
>
> These files are volatile and vulnerable to browser cache clearing or session termination. You can download these files to your local drive by right-clicking a file in the sidebar and selecting *Download*.
>
> **Updates.** A notebook that you have already opened stays as it was when you last opened it, and does not pick up later updates to this site. To get the latest version, delete the notebook in the sidebar and reload the page.
